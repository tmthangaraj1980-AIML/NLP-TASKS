        2. Remove both default and custom stop words from a given document.
        3. compare the results using both NLTK and spacy packages.

##Nltk package

```python
##import natural language toolkit libraries
import nltk
from nltk.corpus import stopwords        ##(stop common English words is, in an etc..)
from nltk.tokenize import word_tokenize  ##(splits sentence into words)
```

```python
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
text = "Natural Language Processing (NLP) is a field of Artificial Intelligence"
```

```python
words = word_tokenize(text)  # split the sentence into words
print('\ntokenized_words:\n', words)
```

```python
stop_words = set(stopwords.words('english')) #Loads default English stop words.
stop_words
```

```python
custom_stopwords = {"field"} #Adds our own stop word., here want to remove word 'field'
```

```python
all_stopwords = stop_words.union(custom_stopwords) #Combines default stop words + custom stop words.
```

```python
filtered_words = []     #fiklter all custom_stop words and default stop words, rest stored as list
for word in words:
    if word.lower() not in all_stopwords:
        filtered_words.append(word)
```

### Output

```python
print(filtered_words)
```

###Simple theory for lab record

Stop words are common words that do not give much meaning in NLP. Removing stop words reduces unnecessary words and keeps only important words.

##Result:
Thus the stop words and custom stop words were successfully removed from the given document using NLTK.

##spaCy packages

```python
# import the spacy library
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")    #Load the small English language model
                                      #This model helps spaCy understand English words, grammar, stop words, POS tags, etc.
```

```python
doc = nlp(text) #The sentence is processed by spaCy.

                #Now doc contains tokens/words with NLP information.
doc
```

```python
useful_words = []  # after the removal of stop_words
for token in doc:
    if not token.is_stop and token.text.lower() not in custom_stopwords:
      useful_words.append(token.text)
```

```python
print(useful_words)
```

##Result:

Thus the default stop words and custom stop words were successfully removed from the given document and the results were compared using both NLTK and spaCy packages.
