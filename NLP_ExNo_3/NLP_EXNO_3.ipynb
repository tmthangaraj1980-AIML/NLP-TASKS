#Implement a Stemming and Lemmatization program

##Aim:
The aim of this program to stemmer and lemmatizer to reduce words to theirbase or root form.

      a. stemming - chopping of prefixes/suffixes to get thge stem (not a valid word)
      b. Lemmatization - converts the word into the dictionary base form(lemma), valid word

```python
#import libraries
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('wordnet')  #WordNet dictionary is for Lemmatizer
```

## Inputs

```python
words = ('playing', 'running', 'studies', 'better', 'jumps')
```

```python
stemmer = PorterStemmer()  #creates stemming object
lemmatizer = WordNetLemmatizer()   # creates lemmatizing object
```

## Output

```python
print('\nsteming and lemmantizing resuls\n')
```

```python
# stemming loop
for word in words:
  print(f'{word} -> {stemmer.stem(word)}')
```

```python
# lemmatizing loop
for word in words:
  print(f'{word} -> {lemmatizer.lemmatize(word, pos='v')}')
```

## Result:

Thus the stemming and lemmatization processes were successfully implemented to reduce words into their root/base forms using NLTK.
